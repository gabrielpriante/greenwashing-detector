```python
from pathlib import Path
import sys

# Locate project root and src/ folder
PROJECT_ROOT = Path("..").resolve()
SRC_DIR = PROJECT_ROOT / "src"
sys.path.append(str(SRC_DIR))

from text_cleaning import basic_clean_text, tokenize
from greenwashing_scoring import simple_greenwashing_score
```

```python
def show_score(text: str):
    result = simple_greenwashing_score(text)
    print("TEXT:")
    print(text)
    print("-" * 60)
    print(f"Score:      {result['score']}")
    print(f"Risk level: {result['risk_level']}")
    print(f"Keywords:   {result['matched_keywords']}")
    print("=" * 60)
    print()
```

```python
examples = [
    # High-risk, fluffy sounding
    "Our eco-friendly product is 100% natural and chemical-free, designed with the planet in mind.",
    
    # Low-risk, concrete target
    "We will reduce our Scope 1 and Scope 2 emissions by 50% by 2030, from a 2019 baseline, verified by a third party.",
    
    # Medium-ish, generic green language
    "This bank is committed to sustainable growth and a greener future for all our stakeholders.",
    
    # Another fluffy one
    "Our clothing line is planet friendly and non toxic, helping you make better choices for the environment.",
]

for text in examples:
    show_score(text)
```
