```python
Markdown("""
## 🎯 Project Purpose

This project provides a reproducible framework for detecting potential greenwashing 
in corporate sustainability text.  
The system combines:

- **Text preprocessing**
- **Keyword-based scoring**
- **Machine learning classification**
- **A unified analysis pipeline**
- **An interactive Streamlit application**

The goal is to create a transparent, extensible, and modular detection tool that can be
expanded with new models, new features, and external data sources.

---
## 🏗️ System Architecture Overview

The detection system includes the following components:

1. **Text Cleaning Module (`src/text_cleaning.py`)**
    - Standardizes case
    - Removes punctuation
    - Normalizes spacing
    - Prepares text for modeling

2. **Keyword Scoring Module (`src/greenwashing_scoring.py`)**
    - Matches against known vague sustainability terms
    - Computes a risk score
    - Flags suspicious language patterns

3. **Machine Learning Model (`models/`)**
    - TF-IDF vectorizer trained on labeled sustainability claims
    - Logistic Regression classifier
    - Identifies greenwashing patterns statistically

4. **Unified Pipeline (`04_unified_pipeline.ipynb`)**
    - Combines cleaning, scoring, and ML prediction
    - Provides both single-text and batch analysis workflows

5. **Streamlit App (`src/app.py`)**
    - Simple user interface for interactive claim analysis
    - Displays keyword matches, model confidence, and risk indicators

---
""")
```

```python
Markdown("""
## 🔄 System Data Flow

This section describes how a piece of input text moves through the system.

1. **Input**
   - Corporate sustainability or ESG claim text.

2. **Text Cleaning (`src/text_cleaning.py`)**
   - Lowercasing
   - Punctuation removal
   - Whitespace normalization

3. **Parallel Processing Paths**
   - **Keyword Scoring (`src/greenwashing_scoring.py`)**
     - Matches against a curated list of greenwashing-related phrases.
     - Produces:
       - A numeric risk score
       - A categorical risk level (Low / Medium / High)
       - A list of matched phrases
   - **Machine Learning Model (`models/`)**
     - TF-IDF vectorizer converts cleaned text to numeric features.
     - Logistic Regression model predicts a greenwashing label and probability.

4. **Unified Pipeline (`04_unified_pipeline.ipynb`)**
   - Combines cleaning, keyword scoring, and model prediction.
   - Returns a structured result containing:
     - Original text
     - Cleaned text
     - Keyword score and matched keywords
     - Model prediction and confidence

5. **Application Layer (`src/app.py`)**
   - Streamlit-based user interface.
   - Accepts text input.
   - Calls the unified pipeline.
   - Displays keyword score, model probability, and a human-readable interpretation.
""")
```

```python
from IPython.display import Markdown

Markdown("""
## 🔗 Unified Pipeline Summary

The unified pipeline combines all major components of the project into a single
analysis function. This function is used by both the Jupyter notebooks and the
Streamlit app.

### Pipeline Steps

1. **Clean the text**  
   Standardizes the input using the cleaning function.

2. **Keyword scoring**  
   Evaluates the text for sustainability-related buzzwords and assigns a score.

3. **TF-IDF vectorization**  
   Converts the cleaned text into numerical features used by the ML model.

4. **Machine learning prediction**  
   Uses a logistic regression classifier to estimate greenwashing likelihood.

5. **Final output**  
   Packages the cleaned text, keyword score, model prediction, and probability
   into a structured result that can be used for display or further processing.

---
""")
```
